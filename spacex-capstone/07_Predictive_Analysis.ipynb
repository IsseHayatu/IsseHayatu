```python
results = json.loads((R/'analysis_results.json').read_text())
cols=['model','cv_accuracy','test_accuracy','precision','recall','f1']
display(pd.DataFrame(results['results'])[cols])
svm = next(row for row in results['results'] if row['model']=='SVM')
display(pd.DataFrame(svm['confusion_matrix'],index=['Actual failure/other','Actual success'],columns=['Predicted failure/other','Predicted success']))
print('Training:',results['train_n'],'Test:',results['test_n'])
print('SVM parameters:',svm['params'])
```

The main comparison uses a StandardScaler pipeline, five-fold stratified validation and a fixed 18-row test set.
