## Per-site launch-record markers
Every one of the 56 records becomes a marker. Color encodes class, the icon shows source flight number, and popups show site/date/outcome. MarkerCluster spiderfy spreads colocated records on screen while preserving source coordinates.

```python
runpy.run_path(str(R/'folium_site_records.py'))
audit=g.groupby('Launch Site')['class'].agg(['count','sum'])
audit['class_0']=audit['count']-audit['sum']
audit['class_1_rate']=audit['sum']/audit['count']
display(audit)
assert audit['count'].sum()==56
assert audit['sum'].sum()==24
display(g[['Launch Site','Flight Number','Date','Landing Outcome','class','Lat','Long']])
```
