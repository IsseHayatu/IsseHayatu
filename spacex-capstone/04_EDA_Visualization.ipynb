Six plots compare outcomes with flight number, launch site, orbit, payload and year. These are historical associations, not causal effects.

```python
from pathlib import Path
import json, pandas as pd
from IPython.display import display
R = Path('.')
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import Image
df = pd.read_csv(R/'data/dataset_part_2.csv')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.scatterplot(data=df,x='FlightNumber',y='LaunchSite',hue='Class')
plt.tight_layout()
plt.savefig('_eda_plot.png', dpi=150)
display(Image(filename='_eda_plot.png'))
plt.close()
```

```python
sns.scatterplot(data=df,x='PayloadMass',y='LaunchSite',hue='Class')
plt.tight_layout()
plt.savefig('_eda_plot.png', dpi=150)
display(Image(filename='_eda_plot.png'))
plt.close()
```

```python
df.groupby('Orbit').Class.mean().plot.bar(ylabel='Landing success rate')
plt.tight_layout()
plt.savefig('_eda_plot.png', dpi=150)
display(Image(filename='_eda_plot.png'))
plt.close()
```

```python
sns.scatterplot(data=df,x='FlightNumber',y='Orbit',hue='Class')
plt.tight_layout()
plt.savefig('_eda_plot.png', dpi=150)
display(Image(filename='_eda_plot.png'))
plt.close()
```

```python
sns.scatterplot(data=df,x='PayloadMass',y='Orbit',hue='Class')
plt.tight_layout()
plt.savefig('_eda_plot.png', dpi=150)
display(Image(filename='_eda_plot.png'))
plt.close()
```

```python
df.assign(Year=pd.to_datetime(df.Date).dt.year).groupby('Year').Class.mean().plot(marker='o',ylabel='Landing success rate')
plt.tight_layout()
plt.savefig('_eda_plot.png', dpi=150)
display(Image(filename='_eda_plot.png'))
plt.close()
```
