```python
from pathlib import Path
import json, pandas as pd
from IPython.display import display
R = Path('.')
import sqlite3
con = sqlite3.connect(':memory:')
df = pd.read_csv(R/'data/Spacex.csv')
for col in df.select_dtypes('object'):
    df[col] = df[col].str.strip()
df.to_sql('SPACEXTBL',con,index=False)
print('SQL rows:',len(df))
```

```python
results = json.loads((R/'analysis_results.json').read_text())
for label, query in results['queries'].items():
    print('\n'+label+'\n'+query)
    display(pd.read_sql_query(query, con))
```
